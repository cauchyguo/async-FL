```python
import os
import re
import argparse

def extract_data_from_log(log_dir):
    """
    从指定目录下的output.log文件中提取准确率、损失值和运行时间信息，
    并生成accuracy.txt, loss.txt和update_time.txt三个文件。
    
    参数:
        log_dir (str): 日志文件所在的目录
    
    返回:
        bool: 是否成功提取数据并生成文件
    """
    log_path = os.path.join(log_dir, 'output.log')
    
    # 检查日志文件是否存在
    if not os.path.exists(log_path):
        print(f"错误: 在{log_dir}目录中未找到output.log文件")
        return False
    
    # 读取日志文件
    try:
        with open(log_path, 'r', encoding='utf-8') as f:
            log_content = f.read()
    except Exception as e:
        print(f"读取日志文件时出错: {e}")
        return False
    
    # 提取测试结果
    accuracy_data = []
    loss_data = []
    time_data = []
    
    # 匹配测试结果的正则表达式
    # 匹配形如 "Epoch 1 tested, accuracy: 14.24... loss 2.75... run_time 37.43..." 的行
    test_result_regex = r"Epoch (\d+) tested, accuracy: ([\d.]+) loss ([\d.]+) run_time ([\d.]+)"
    matches = re.findall(test_result_regex, log_content)
    
    if not matches:
        print("警告: 未在日志中找到测试结果数据")
        return False
    
    # 按轮次排序，确保数据顺序正确
    matches.sort(key=lambda x: int(x[0]))
    
    for match in matches:
        epoch = int(match[0])
        accuracy = float(match[1])
        loss = float(match[2])
        run_time = float(match[3])
        
        accuracy_data.append(accuracy)
        loss_data.append(loss)
        time_data.append(run_time)
        
        print(f"提取到第{epoch}轮数据: 准确率={accuracy:.4f}, 损失={loss:.4f}, 时间={run_time:.4f}")
    
    # 写入提取的数据到文件
    try:
        # 准确率文件
        with open(os.path.join(log_dir, 'accuracy.txt'), 'w') as f:
            f.write(str(accuracy_data))
        
        # 损失值文件
        with open(os.path.join(log_dir, 'loss.txt'), 'w') as f:
            f.write(str(loss_data))
        
        # 运行时间文件
        with open(os.path.join(log_dir, 'update_time.txt'), 'w') as f:
            f.write(str(time_data))
        
        print(f"\n成功提取数据并生成文件!")
        print(f"- 共提取了{len(accuracy_data)}轮数据")
        print(f"- 文件保存在: {log_dir}")
        return True
    
    except Exception as e:
        print(f"写入文件时出错: {e}")
        return False

def main():
    # 命令行参数解析
    log_dir = "/disk6T/ypguo/async-FL/PK/Exp2025-V2-Cifar10-dir0.1/FedBuff-dir0.1"    
    # 提取数据
    extract_data_from_log(log_dir)

if __name__ == "__main__":
    main()
```
